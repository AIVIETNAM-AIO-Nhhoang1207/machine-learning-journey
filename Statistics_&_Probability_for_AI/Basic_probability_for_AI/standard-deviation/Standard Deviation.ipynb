```python
import numpy as np

x = np.array([5, 3, 6, 7, 4])

var = (x - x.mean())**2
var = np.mean(var)
print(var)

sdv = np.sqrt(var)
print(sdv)
```

```python
import cv2
import numpy as np
from scipy.ndimage import generic_filter

img_gray = cv2.imread("img.jpg", cv2.IMREAD_GRAYSCALE)

cv2.imwrite("img_gray.jpg", img_gray)

x = img_gray.astype(np.float32)

texture = generic_filter(x, np.std, size=7)

texture = cv2.normalize(
    texture,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)

texture = texture.astype(np.uint8)

cv2.imwrite("texture.jpg", texture)
```

```python
import numpy as np
import cv2
import math
from scipy.ndimage import generic_filter


img  = cv2.imread('img.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite('edge_s1.jpg', gray)

x = gray.astype('float')
x_filt = generic_filter(x, np.std, size=7)
cv2.imwrite('edge_s2.jpg', x_filt)

x_filt[x_filt < 20] = 0
cv2.imwrite('edge_s3.jpg', x_filt)

maxv = np.max(x_filt)
print(maxv)

x_filt = x_filt*2.5
cv2.imwrite('edge_s4.jpg', x_filt)
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from matplotlib import cm

img = cv2.imread("img.jpg", cv2.IMREAD_GRAYSCALE)

# ROI
img = img[0:50, 0:50]

# Tạo tọa độ
xx, yy = np.mgrid[0:img.shape[0], 0:img.shape[1]]

# Figure
fig = plt.figure(figsize=(8,6))

# Thay gca() bằng add_subplot()
ax = fig.add_subplot(111, projection='3d')

# Vẽ mặt
ax.plot_surface(
    xx,
    yy,
    img,
    cmap=cm.coolwarm,
    rstride=1,
    cstride=1,
    linewidth=0,
    antialiased=True
)

ax.set_zlim(0,255)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Intensity")

plt.show()
```
